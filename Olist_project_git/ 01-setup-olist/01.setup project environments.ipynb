# 01.Setup the project environment for formula 1
- Create new extrenal location for your container formula 1
- Create new catalog for formula1 project

External Location

```python
%sql
create external location if not exists databricks_ext_9860_formula1_incr
url 'abfss://demo@databricksextdl9860.dfs.core.windows.net/'
with (storage credential `databricks-course-sc`)
comment 'External location for the Olist  container';
```

Access cloud containers storage

New Catalog for Formula1

```python
%sql
show catalogs;
```

```python
%sql
create catalog if not exists ecommerce_olist
managed location 'abfss://demo@databricksextdl9860.dfs.core.windows.net/'
comment 'this is a catalog for the olist  database'
;
```

Create Schema/Database for all bronze gold silver layer

```python
%sql
create schema if not exists ecommerce_olist.landing;
create schema if not exists ecommerce_olist.bronze
    managed location 'abfss://demo@databricksextdl9860.dfs.core.windows.net/bronze';

create schema if not exists ecommerce_olist.silver
    managed location 'abfss://demo@databricksextdl9860.dfs.core.windows.net/silver';

create schema if not exists ecommerce_olist.gold
    managed location 'abfss://demo@databricksextdl9860.dfs.core.windows.net/gold'
```
